# PZT atomiser final analysis

The PZT calibration is 5.35 micrometres/pixel with pulse separation
25 microseconds: instantaneous pixel displacements are converted once by
0.214 m/s. This calibration supersedes both 5.05 and 7.25.
Cartesian x/u is horizontal/transverse; y/v is vertical/axial.
Naming changes do not transpose, rotate or reflect the field. The exported
image-reference origin and existing station indices/frame keys are retained.

Mean and instantaneous axial contours, including the station map, display
all finite signed measurements. Blue is negative, white is zero and red is
positive. Symmetric limits and one linear normalization are used, with
uniformly spaced contour levels and no separate stretching of signs. Each
mean component uses its own maximum absolute ensemble value, excluding flag
diagnostics. Selected instantaneous fields share limits. Only missing or
invalid data are masked. Values are recalculated from the original displacement
samples; sample selection, station indices, averaging and orientation are retained.

## Proposed side-jet combined-flow prediction

PZT-only measurements supply droplet velocities; the surrounding air was not
measured and the jet was absent. ZNMF-only measurements supply a separate gas
field. The combined experiment has not been conducted. Baselines are generated
without assuming alignment. The configured nominal diameter is a model estimate,
not a measurement: 5 micrometre aperture, assumed equal jet diameter, Rayleigh
factor 1.89 giving 9.45 micrometres, rounded to 10 micrometres for all calculations.
Diameter remains configurable. Predictions still require nozzle placement and
PZT release nodes in config/combined_flow_prediction.json.
The solver uses evolving vector slip in a prescribed measured mean gas field;
PZT velocities are used once at initialisation, never added to the gas field.


```python
from PZT_Atomiser_Analysis.analysis.combined_flow_prediction import run as run_prediction
prediction=run_prediction()
print(json.dumps(prediction,indent=2))
package=output/'Combined_Flow_Prediction'
manifest=json.loads((package/'report_manifest.json').read_text())
for figure in manifest['figures']:
    print(figure['kind'],figure['stem'])
    display(Image(filename=str(package/'figures'/(figure['stem']+'.png')),width=650))
```
